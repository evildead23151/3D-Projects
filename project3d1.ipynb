```python
# =========================
# 1. Install Open3D (run once per session)
# =========================
!pip install open3d

# =========================
# 2. Imports
# =========================
import numpy as np
import open3d as o3d

# =========================
# 3. Simulated sonar parameters
# =========================

# Angular fan of sonar rays (±30 degrees)
angles = np.linspace(-np.pi / 6, np.pi / 6, 200)

# Target surface: plane x = 5 meters
plane_x = 5.0

points = []

# =========================
# 4. Ray → surface intersection
# =========================
for theta in angles:
    # Ray direction (2D fan in XY plane)
    dx = np.cos(theta)
    dy = np.sin(theta)

    # Avoid division explosion (grazing rays)
    if abs(dx) < 1e-6:
        continue

    # Intersection with plane x = 5
    t = plane_x / dx

    # Cartesian coordinates
    x = t * dx
    y = t * dy
    z = 0.0  # Flat surface assumption

    points.append([x, y, z])

points = np.array(points)

# =========================
# 5. Create Open3D Point Cloud
# =========================
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)

# Optional: color for clarity
pcd.paint_uniform_color([0.2, 0.8, 1.0])

# =========================
# 6. Visualize
# =========================
o3d.visualization.draw_geometries(
    [pcd],
    window_name="Sonar Ray → Point Cloud",
    width=900,
    height=600
)
```

```python
o3d.io.write_point_cloud("sonar_plane.ply", pcd)
print("Saved sonar_plane.ply")
```

```python
import plotly.graph_objects as go

points_np = np.asarray(pcd.points)

fig = go.Figure(
    data=[
        go.Scatter3d(
            x=points_np[:, 0],
            y=points_np[:, 1],
            z=points_np[:, 2],
            mode='markers',
            marker=dict(size=3)
        )
    ]
)

fig.update_layout(
    scene=dict(
        xaxis_title="X",
        yaxis_title="Y",
        zaxis_title="Z",
        aspectmode="data"
    ),
    title="Sonar Ray → Point Cloud (Headless View)"
)

fig.show()
```
